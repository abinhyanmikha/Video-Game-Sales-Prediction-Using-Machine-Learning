Identify potential outlier using IQR

```python
cols = [
    "Global_Sales",
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Critic_Score",
    "Critic_Count",
    "User_Count"
]

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\n{col}")
    print("Potential outliers:", len(outliers))
    print("Lower bound:", round(lower, 2))
    print("Upper bound:", round(upper, 2))
```

```python
# Check for negative sales
sales_cols = [
    "NA_Sales",
    "EU_Sales",
    "JP_Sales",
    "Other_Sales",
    "Global_Sales"
]

for col in sales_cols:
    print(f"{col}: {(df[col] < 0).sum()} negative values")

# Check critic score range
print("Critic score range:")
print(df["Critic_Score"].min(), df["Critic_Score"].max())

# Check release year range
print("Release year range:")
print(df["Year_of_Release"].min(), df["Year_of_Release"].max())
```
